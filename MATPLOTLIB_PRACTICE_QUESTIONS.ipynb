```python
import matplotlib.pyplot as plt
import numpy as np
products = ["Laptop", "Mobile", "Tablet", "Headphones", "Smartwatch"]
product_sales = [85000, 120000, 65000, 45000, 55000]

departments = ["HR", "IT", "Sales", "Finance", "Marketing"]
employees = [12, 35, 28, 15, 22]

categories = ["Electronics", "Clothing", "Furniture", "Grocery"]
category_sales = [180000, 120000, 90000, 160000]

age_groups = ["18-25", "26-35", "36-45", "46-55", "56+"]
customers = [120, 250, 180, 100, 50]
```

```python
plt.bar(products,product_sales,label="product_sales")
plt.title("product_sales")
plt.xlabel("Sales")
plt.ylabel("Products")
plt.legend()
plt.show()
```

```python
plt.plot(products,product_sales,label="product_sales", marker="o")
plt.title("product_sales")
plt.xlabel("Sales")
plt.ylabel("Products")
plt.legend()
plt.show()
```

```python
plt.barh(products,product_sales,label="product_sales")
plt.title("product_sales")
plt.xlabel("Sales")
plt.ylabel("Products")
plt.legend()
plt.show()
```

```python
plt.pie(
    product_sales,
    labels=products,
    autopct="%1.1f%%",
    wedgeprops={"width":0.6}
)
plt.title(product_sales)
plt.show()
```
